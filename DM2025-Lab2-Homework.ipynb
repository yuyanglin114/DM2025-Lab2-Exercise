# **Project Report**

## **1. Model Development (Final Model)**
### **1.1 Preprocessing Steps**

The preprocessing stage focused on constructing a clean and reliable text–label dataset from the provided Kaggle competition files — `emotion.csv`, `data_identification.csv`, and `samplesubmission.csv` — together with the textual content extracted from the `final_posts.json` Twitter archive.

The key steps were:

1. **Data Integration**

    - `data_identification.csv` was used to determine the train/test split for each `id`.

    - `emotion.csv` provided emotion labels for training instances.

    - Text data were loaded from the JSON tweet archive and merged using the `id` field.

    - Since Kaggle files do not contain text directly, the tweet JSON served as the authoritative source.

2. **Cleaning and Standardization**

    - All text fields were converted to strings.

    - Missing values were replaced with empty strings (`""`) to prevent vectorization or tokenization errors.

    - Light normalization was applied: lowercasing, trimming whitespace, and removing malformed characters.

    - Duplicate tweet IDs (rare, due to merging) were removed.

3. **Train–Validation Split**
    - The training portion was stratified into **80% train / 20% validation** to preserve label distribution.

    - This validation set served both for model selection and for monitoring overfitting.

**The final dataset consisted of:**
- **Train:** 38,312 tweets

- **Validation:** 9,578 tweets

- **Test:** unlabeled tweets for Kaggle submission

**Each record included:**
- id, text, emotion (train only), split

---

### **1.2 Feature Engineering Steps**

Even though transformer models process raw text directly, several feature engineering decisions significantly improved model performance.

1. **Tokenization**

    The `distilbert-base-uncased` tokenizer was used with the following configuration:

    - Lowercased vocabulary

    - `max_length = 160` tokens

    - Padding to max length

    - Truncation when sequences exceeded the maximum

    Tokenization produced:

    - `input_ids`

    - `attention_mask`

    These were fed into the transformer encoder.

2. **Label Encoding**

    Emotion categories
    - `anger, disgust, fear, joy, sadness, surprise`
    were encoded using `LabelEncoder`, mapping each to an integer.

3. **DataLoader Construction**

    PyTorch `Dataset` and `DataLoader` objects were created to:

    - Batch input efficiently

    - Shuffle training samples

    - Enable GPU-accelerated training

4. **Checkpointing**

    After each epoch:

    - The model was evaluated on the validation set.

    - The checkpoint with the best **macro F1** score was saved.

    - The final submission used this best-performing model rather than the last-epoch weights.

    This ensured robustness and mitigated overfitting.

---

### **1.3 Explanation of the Model**

The final model is a **DistilBERT-based emotion classifier**, chosen because it offers a strong balance between accuracy and computational efficiency.

#### **Architecture Overview**

1. **DistilBERT Encoder**

    - 6-layer Transformer architecture (a distilled version of BERT-base)

    - Generates contextual embeddings for each token

    - Faster and lighter (~40% fewer parameters) but still strong on semantic tasks

2. **Classification Layer**

    - Linear layer applied to the `[CLS]` embedding

    - Outputs logits over 6 emotion categories

    - Softmax applied during inference

#### **Training Configuration**

- **Loss:** Cross-entropy (final model uses no class weights)

- **Optimizer:** AdamW (`LR = 2e-5`)

- **Scheduler:** Linear warmup + linear decay

- **Epochs:** 3

- **Batch size:** 16 (train) / 32 (validation & test)

- **Device:** GPU when available

#### **Why This Model Was Chosen**

Experiments showed:

- Models with heavy class weighting increased validation macro F1 but generalized poorly on Kaggle.

- The best leaderboard score came from **a stable, unweighted DistilBERT model trained for multiple epochs with checkpointing.**

Final leaderboard performance:

    - Kaggle Public Score: 0.6809 (60/153)
    - Kaggle Private Score: 0.6649 (65/153)

This confirms the final model is both effective and reliable.

### **2.1 Different Things I Tried**

Several modeling strategies were explored:

#### **1. TF-IDF + Linear Models**

- Logistic Regression & Linear SVM baselines

- Provided quick sanity checks

- Much weaker than transformer models → justified moving to BERT-based approaches

### **2. DistilBERT with Class Weights**

- Applied inverse-frequency weights to correct label imbalance

- Boosted recall for minority classes (e.g., disgust)

- Severely hurt precision and overall generalization

- Resulted in a **lower** Kaggle leaderboard score

### **3. Multi-Epoch Training + Best Checkpoint**

- Macro F1 steadily increased across epochs

- Validation and Kaggle performance improved significantly

- Ultimately chosen as part of the final model

### **4. Sequence Length Experiments**

- Tried 128, 160, and 192 tokens

- Longer sequences improved contextual understanding slightly

`max_length = 160` gave the best tradeoff

### **5. Larger Model Exploration (BERT-base)**

- Considered upgrading to `bert-base-uncased`

- Preliminary tests promising, but omitted due to computational cost

---

## **2.2 Insights I Gained**
### **1. Macro F1 vs. Real-World Generalization**

Increasing macro F1 (via class weighting) does not guarantee better leaderboard performance.
Leaderboard evaluation favored models that maintained stable predictions on larger classes.

### **2. Stability > Forced Balance**

Overly correcting minority classes caused the model to misclassify many majority-class examples.

### **3. DistilBERT Is Extremely Competitive**

DistilBERT achieved excellent performance with far lower computational cost than full BERT.

### **4. Checkpointing Is Crucial**

The best-performing model was not the final epoch.
Saving the best validation checkpoint prevented unnecessary degradation.

### **5. Kaggle Behavior Reflects Real Deployments**

The test set behaved more like a naturally imbalanced dataset, confirming that:

    - Balanced accuracy does not always mean better real-world performance.

# **Competition Code**

## 1. Preprocessing Steps

```python
### 1. Preprocessing Steps

import os
import json
import pandas as pd

# --- 路徑設定 ---
BASE_DIR = r"D:\GitHub\Data Mining\DM2025Labs\DM2025-Lab2-Exercise"
COMP_DIR = os.path.join(BASE_DIR, "dm-lab-2-private-competition")

EMOTION_PATH    = os.path.join(COMP_DIR, "emotion.csv")
DATA_IDENT_PATH = os.path.join(COMP_DIR, "data_identification.csv")
SAMPLE_SUB_PATH = os.path.join(COMP_DIR, "samplesubmission.csv")
POSTS_JSON_PATH = os.path.join(COMP_DIR, "final_posts.json") 

print("COMP_DIR:", COMP_DIR)
print("POSTS_JSON_PATH:", POSTS_JSON_PATH)

# 1-1. 讀入 Kaggle 競賽基本檔案
emotion_df     = pd.read_csv(EMOTION_PATH)       # id, emotion（只有 train 有）
data_ident_df  = pd.read_csv(DATA_IDENT_PATH)    # id, split (train/test)
sample_sub_df  = pd.read_csv(SAMPLE_SUB_PATH)    # id, emotion（提交格式範例）

print("emotion_df:", emotion_df.shape)
print("data_ident_df:", data_ident_df.shape)
print("sample_sub_df:", sample_sub_df.shape)

# 1-2. 讀入 tweet 文字：final_posts.json
with open(POSTS_JSON_PATH, "r", encoding="utf-8") as f:
    posts_raw = json.load(f)

posts = []
for item in posts_raw:
    post = item["root"]["_source"]["post"]
    posts.append({
        "id": post["post_id"],
        "text": post["text"]
    })

text_df = pd.DataFrame(posts)
print("text_df:", text_df.shape)
print(text_df.head())

# 1-3. 型別統一，以免後面 merge 出現 int / object 衝突
emotion_df["id"]    = emotion_df["id"].astype(str)
data_ident_df["id"] = data_ident_df["id"].astype(str)
text_df["id"]       = text_df["id"].astype(str)

# 1-4. 合併 split + label + text
full_df = data_ident_df.merge(emotion_df, on="id", how="left")
full_df = full_df.merge(text_df, on="id", how="left")

# 補缺值，確保 text 是字串
full_df["text"] = full_df["text"].fillna("").astype(str)

print(full_df.head())
print("split counts:\n", full_df["split"].value_counts())
print("empty text rows:", (full_df["text"].str.len() == 0).sum())

# 1-5. 切成 train / test 供後續使用
train_df = full_df[full_df["split"] == "train"].copy()
test_df  = full_df[full_df["split"] == "test"].copy()

print("Train size:", train_df.shape)
print("Test  size:", test_df.shape)
print(train_df[["id", "emotion", "text"]].head())
```

## 2. Feature Engineering Steps

```python
### 2. Feature Engineering Steps

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

# 2-1. 準備文字(X)與標籤(y)
X_text = train_df["text"].astype(str)
y      = train_df["emotion"]

# 2-2. 切出 validation set（用來評估傳統模型與 BERT）
X_train_text, X_valid_text, y_train, y_valid = train_test_split(
    X_text,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Train samples:", len(X_train_text))
print("Valid samples:", len(X_valid_text))

# 2-3. 使用 TF-IDF 作為 baseline 的文字特徵
vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),  # unigram + bigram
    min_df=2,
    max_df=0.98
)

X_train_vec = vectorizer.fit_transform(X_train_text)
X_valid_vec = vectorizer.transform(X_valid_text)

print("X_train_vec shape:", X_train_vec.shape)
print("X_valid_vec shape:", X_valid_vec.shape)

# 後面 Model 寫法會用到的完整 train/test 特徵
X_full_vec = vectorizer.fit_transform(train_df["text"].astype(str))
X_test_vec = vectorizer.transform(test_df["text"].astype(str))
```

## 3. Model Implementation Steps

```python
### (v3). DistilBERT without class weights + best checkpoint

import torch
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
import torch.nn.functional as F

from transformers import AutoTokenizer, AutoModelForSequenceClassification
from transformers.optimization import get_linear_schedule_with_warmup

from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, f1_score
from tqdm.auto import tqdm
import numpy as np
import pandas as pd

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

MODEL_NAME = "distilbert-base-uncased"
tokenizer  = AutoTokenizer.from_pretrained(MODEL_NAME)

# === 1. Label encoding ===
le = LabelEncoder()
y_train_ids = le.fit_transform(y_train)
y_valid_ids = le.transform(y_valid)

num_labels = len(le.classes_)
print("num_labels:", num_labels, le.classes_)

# === 2. Dataset & DataLoader ===
class TextDataset(Dataset):
    def __init__(self, texts, labels=None, max_len=160):
        self.texts = list(texts)
        self.labels = labels
        self.max_len = max_len

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = str(self.texts[idx])
        enc = tokenizer(
            text,
            truncation=True,
            padding="max_length",
            max_length=self.max_len,
            return_tensors="pt"
        )
        item = {k: v.squeeze(0) for k, v in enc.items()}
        if self.labels is not None:
            item["labels"] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item

train_ds = TextDataset(X_train_text, y_train_ids)
valid_ds = TextDataset(X_valid_text, y_valid_ids)

train_loader = DataLoader(train_ds, batch_size=16, shuffle=True)
valid_loader = DataLoader(valid_ds, batch_size=32)

# === 3. Model, optimizer, scheduler ===
bert_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=num_labels
).to(device)

optimizer = AdamW(bert_model.parameters(), lr=2e-5)

num_epochs = 3
num_training_steps = num_epochs * len(train_loader)
scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * num_training_steps),
    num_training_steps=num_training_steps
)

best_f1 = -1.0
best_state_dict = None

# === 4. Training loop (no class weights) ===
for epoch in range(num_epochs):
    bert_model.train()
    total_loss = 0.0

    for batch in tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs} - train"):
        labels = batch.pop("labels").to(device)
        batch  = {k: v.to(device) for k, v in batch.items()}

        outputs = bert_model(**batch)
        logits  = outputs.logits

        # ❗ 不使用 class_weights，回到普通 cross entropy
        loss = F.cross_entropy(logits, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        scheduler.step()

        total_loss += loss.item()

    avg_train_loss = total_loss / len(train_loader)

    # === Validation ===
    bert_model.eval()
    preds_all, labels_all = [], []
    with torch.no_grad():
        for batch in tqdm(valid_loader, desc=f"Epoch {epoch+1}/{num_epochs} - valid"):
            labels = batch.pop("labels").to(device)
            batch  = {k: v.to(device) for k, v in batch.items()}

            outputs = bert_model(**batch)
            logits  = outputs.logits
            preds   = logits.argmax(dim=-1)

            preds_all.extend(preds.cpu().numpy())
            labels_all.extend(labels.cpu().numpy())

    val_f1 = f1_score(labels_all, preds_all, average="macro")
    print(f"\n[BERT v3] Epoch {epoch+1} train_loss={avg_train_loss:.4f}, valid_macroF1={val_f1:.4f}")

    print("\n[BERT v3] Classification report on validation set:")
    print(
        classification_report(
            labels_all,
            preds_all,
            target_names=le.classes_
        )
    )

    if val_f1 > best_f1:
        best_f1 = val_f1
        best_state_dict = {k: v.cpu().clone() for k, v in bert_model.state_dict().items()}
        print(f"--> [v3] New best model saved (macro F1 = {best_f1:.4f})")

print(f"\n[BERT v3] Training finished. Best validation macro F1 = {best_f1:.4f}")

# === 5. 用 best model 產生 Kaggle submission ===
if best_state_dict is not None:
    bert_model.load_state_dict(best_state_dict)
    bert_model.to(device)

test_ds = TextDataset(test_df["text"].astype(str), labels=None)
test_loader = DataLoader(test_ds, batch_size=32)

bert_model.eval()
test_preds_ids = []
with torch.no_grad():
    for batch in tqdm(test_loader, desc="Predict test with best BERT v3"):
        batch = {k: v.to(device) for k, v in batch.items()}
        outputs = bert_model(**batch)
        logits  = outputs.logits
        preds   = logits.argmax(dim=-1)
        test_preds_ids.extend(preds.cpu().numpy())

test_emotions = le.inverse_transform(test_preds_ids)

submission_bert_v3 = pd.DataFrame({
    "id": test_df["id"],
    "emotion": test_emotions
})

submission_bert_v3 = submission_bert_v3[sample_sub_df.columns]
submission_bert_v3_filename = "submission_bert_distilbert_v3_unweighted.csv"
submission_bert_v3.to_csv(submission_bert_v3_filename, index=False)

print(f"\n[BERT v3] Submission saved as: {submission_bert_v3_filename}")
print(submission_bert_v3.head())
```

### 2. BERT (DistilBERT)

```python
### BERT (DistilBERT) for Emotion Classification

import torch
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
)
from transformers.optimization import get_linear_schedule_with_warmup  # ← 從 optimization 匯入

from sklearn.preprocessing import LabelEncoder
from tqdm.auto import tqdm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

MODEL_NAME = "distilbert-base-uncased"
tokenizer  = AutoTokenizer.from_pretrained(MODEL_NAME)

# 1. 把 label 轉成 0~num_classes-1
le = LabelEncoder()
y_train_ids = le.fit_transform(y_train)
y_valid_ids = le.transform(y_valid)

num_labels = len(le.classes_)
print("num_labels:", num_labels, le.classes_)

class TextDataset(Dataset):
    def __init__(self, texts, labels=None, max_len=128):
        self.texts = list(texts)
        self.labels = labels
        self.max_len = max_len

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = str(self.texts[idx])
        enc = tokenizer(
            text,
            truncation=True,
            padding="max_length",
            max_length=self.max_len,
            return_tensors="pt"
        )
        item = {k: v.squeeze(0) for k, v in enc.items()}
        if self.labels is not None:
            item["labels"] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item

# X_train_text, X_valid_text, y_train, y_valid 來自 Feature Engineering 那一格
train_ds = TextDataset(X_train_text, y_train_ids)
valid_ds = TextDataset(X_valid_text, y_valid_ids)

train_loader = DataLoader(train_ds, batch_size=16, shuffle=True)
valid_loader = DataLoader(valid_ds, batch_size=32)

# 2. 建立 BERT 分類模型
bert_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=num_labels
).to(device)

optimizer = AdamW(bert_model.parameters(), lr=2e-5)

num_epochs = 1
num_training_steps = num_epochs * len(train_loader)
scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * num_training_steps),
    num_training_steps=num_training_steps
)

# 3. 訓練迴圈（簡化版）
for epoch in range(num_epochs):
    bert_model.train()
    total_loss = 0
    for batch in tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs} - train"):
        batch = {k: v.to(device) for k, v in batch.items()}
        outputs = bert_model(**batch)
        loss = outputs.loss

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        scheduler.step()

        total_loss += loss.item()

    avg_train_loss = total_loss / len(train_loader)

    # Validation
    bert_model.eval()
    preds_all, labels_all = [], []
    with torch.no_grad():
        for batch in tqdm(valid_loader, desc=f"Epoch {epoch+1}/{num_epochs} - valid"):
            labels = batch.pop("labels")
            batch  = {k: v.to(device) for k, v in batch.items()}
            outputs = bert_model(**batch)
            logits  = outputs.logits
            preds   = logits.argmax(dim=-1).cpu()

            preds_all.extend(preds.numpy())
            labels_all.extend(labels.numpy())

    from sklearn.metrics import classification_report  # 檔案最上面也可以集中匯入

    bert_f1 = f1_score(labels_all, preds_all, average="macro")
    print(f"[BERT] Epoch {epoch+1} train_loss={avg_train_loss:.4f}, valid_macroF1={bert_f1:.4f}")

    # 產生更詳細的分類報告
    print("\n[BERT] Classification report on validation set:")
    print(
        classification_report(
            labels_all,
            preds_all,
            target_names=le.classes_   # le 是前面 LabelEncoder()，classes_ 就是原本的 emotion 名稱
        )
    )


# 4. 用訓練好的 BERT 對 test_df 做推論並產生 submission

test_ds = TextDataset(test_df["text"].astype(str), labels=None)
test_loader = DataLoader(test_ds, batch_size=32)

bert_model.eval()
test_preds_ids = []
with torch.no_grad():
    for batch in tqdm(test_loader, desc="Predict test with BERT"):
        batch = {k: v.to(device) for k, v in batch.items()}
        outputs = bert_model(**batch)
        logits  = outputs.logits
        preds   = logits.argmax(dim=-1).cpu()
        test_preds_ids.extend(preds.numpy())

test_emotions = le.inverse_transform(test_preds_ids)

submission_bert = pd.DataFrame({
    "id": test_df["id"],
    "emotion": test_emotions
})

submission_bert = submission_bert[sample_sub_df.columns]
submission_bert_filename = "submission_bert_distilbert.csv"
submission_bert.to_csv(submission_bert_filename, index=False)

print(f"\n[BERT] Submission saved as: {submission_bert_filename}")
print(submission_bert.head())
```

### 3. Optimized BERT: class weights + 3 epochs best checkpoint

```python
### (Optimized). BERT (DistilBERT) with class weights + best checkpoint

import torch
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
import torch.nn.functional as F

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
)
from transformers.optimization import get_linear_schedule_with_warmup

from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, f1_score
from tqdm.auto import tqdm
import numpy as np
import pandas as pd

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

MODEL_NAME = "distilbert-base-uncased"  # 若 GPU OK，可試試 "bert-base-uncased"
tokenizer  = AutoTokenizer.from_pretrained(MODEL_NAME)

# === 1. Label encoding ===
le = LabelEncoder()
y_train_ids = le.fit_transform(y_train)
y_valid_ids = le.transform(y_valid)

num_labels = len(le.classes_)
print("num_labels:", num_labels, le.classes_)

# === 2. Compute class weights (handle imbalance) ===
class_counts = np.bincount(y_train_ids, minlength=num_labels)
class_weights = len(y_train_ids) / (num_labels * class_counts)
class_weights = torch.tensor(class_weights, dtype=torch.float32).to(device)

print("class_counts:", class_counts)
print("class_weights:", class_weights.detach().cpu().numpy())

# === 3. Dataset & DataLoader ===
class TextDataset(Dataset):
    def __init__(self, texts, labels=None, max_len=160):
        self.texts = list(texts)
        self.labels = labels
        self.max_len = max_len

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = str(self.texts[idx])
        enc = tokenizer(
            text,
            truncation=True,
            padding="max_length",
            max_length=self.max_len,
            return_tensors="pt"
        )
        item = {k: v.squeeze(0) for k, v in enc.items()}
        if self.labels is not None:
            item["labels"] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item

train_ds = TextDataset(X_train_text, y_train_ids)
valid_ds = TextDataset(X_valid_text, y_valid_ids)

train_loader = DataLoader(train_ds, batch_size=16, shuffle=True)
valid_loader = DataLoader(valid_ds, batch_size=32)

# === 4. Build model, optimizer, scheduler ===
bert_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=num_labels
).to(device)

optimizer = AdamW(bert_model.parameters(), lr=2e-5)

num_epochs = 3  # 建議 3，若時間不夠先設 2
num_training_steps = num_epochs * len(train_loader)
scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * num_training_steps),
    num_training_steps=num_training_steps
)

best_f1 = -1.0
best_state_dict = None

# === 5. Training loop with class-weighted loss & best checkpoint ===
for epoch in range(num_epochs):
    bert_model.train()
    total_loss = 0.0

    for batch in tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs} - train"):
        labels = batch.pop("labels").to(device)
        batch  = {k: v.to(device) for k, v in batch.items()}

        outputs = bert_model(**batch)
        logits  = outputs.logits

        # 使用 class_weights 的 cross entropy
        loss = F.cross_entropy(logits, labels, weight=class_weights)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        scheduler.step()

        total_loss += loss.item()

    avg_train_loss = total_loss / len(train_loader)

    # === Validation ===
    bert_model.eval()
    preds_all, labels_all = [], []
    with torch.no_grad():
        for batch in tqdm(valid_loader, desc=f"Epoch {epoch+1}/{num_epochs} - valid"):
            labels = batch.pop("labels").to(device)
            batch  = {k: v.to(device) for k, v in batch.items()}

            outputs = bert_model(**batch)
            logits  = outputs.logits
            preds   = logits.argmax(dim=-1)

            preds_all.extend(preds.cpu().numpy())
            labels_all.extend(labels.cpu().numpy())

    # Macro F1 on validation
    val_f1 = f1_score(labels_all, preds_all, average="macro")
    print(f"\n[BERT] Epoch {epoch+1} train_loss={avg_train_loss:.4f}, valid_macroF1={val_f1:.4f}")

    print("\n[BERT] Classification report on validation set:")
    print(
        classification_report(
            labels_all,
            preds_all,
            target_names=le.classes_
        )
    )

    # 更新 best checkpoint
    if val_f1 > best_f1:
        best_f1 = val_f1
        best_state_dict = {k: v.cpu().clone() for k, v in bert_model.state_dict().items()}
        print(f"--> New best model saved (macro F1 = {best_f1:.4f})")

print(f"\nTraining finished. Best validation macro F1 = {best_f1:.4f}")

# === 6. 用 best model 在 test 上推論 & 產生 submission ===
if best_state_dict is not None:
    bert_model.load_state_dict(best_state_dict)
    bert_model.to(device)

test_ds = TextDataset(test_df["text"].astype(str), labels=None)
test_loader = DataLoader(test_ds, batch_size=32)

bert_model.eval()
test_preds_ids = []
with torch.no_grad():
    for batch in tqdm(test_loader, desc="Predict test with best BERT"):
        batch = {k: v.to(device) for k, v in batch.items()}
        outputs = bert_model(**batch)
        logits  = outputs.logits
        preds   = logits.argmax(dim=-1)
        test_preds_ids.extend(preds.cpu().numpy())

test_emotions = le.inverse_transform(test_preds_ids)

submission_bert = pd.DataFrame({
    "id": test_df["id"],
    "emotion": test_emotions
})

submission_bert = submission_bert[sample_sub_df.columns]
submission_bert_filename = "submission_bert_distilbert_v2_weighted.csv"
submission_bert.to_csv(submission_bert_filename, index=False)

print(f"\n[BERT v2] Submission saved as: {submission_bert_filename}")
print(submission_bert.head())
```
